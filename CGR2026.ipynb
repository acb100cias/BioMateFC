```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import uniform
```

```python
def cgr(dna):
  dna_x=dna[0]
  dna_y=dna[1]
  lenght= len(dna_x)
  x=[uniform.rvs()]
  y=[uniform.rvs()]
  for i in range(lenght):
    x.append(x[i]+0.5*(dna_x[i]-x[i]))
    y.append(y[i]+0.5*(dna_y[i]-y[i]))
  return [x,y]

def traduce(dna):
  dna_x=[]
  dna_y=[]
  for i in range(len(dna)):
    if dna[i]=='A':
      dna_x.append(0)
      dna_y.append(0)
    elif dna[i]=='C':
      dna_x.append(0)
      dna_y.append(1)
    elif dna[i]=='G':
      dna_x.append(1)
      dna_y.append(0)
    elif dna[i]=='T':
      dna_x.append(1)
      dna_y.append(1)
  return [dna_x,dna_y]
```

```python
def main(nombre):
  dna=[]
  archivo=open(nombre)
  read=archivo.read()
  for i in read:
    el=i.upper()
    if el == 'A' or el =='C' or el=='G' or el=='T':
      dna.append(el)
    elif el=='N':
      print('FIN')
  data=cgr(traduce(dna))
  plt.plot(data[0],data[1],'.k',markersize=2)
  plt.savefig('test.png')
  plt.show()


```
