```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as itg
```

```python
def SIR(y,t,alpha,gamma):
  S,I,R=y
  dS=-alpha*S*I/N
  dI=alpha*S*I/N-gamma*I
  dR=gamma*I
  return [dS,dI,dR]

```

```python
def plotSol(S,I,R,factor=1):
  fig,ax=plt.subplots(1,1)
  ax.plot(t,S/factor,label='S')
  ax.plot(t,I/factor,label='I')
  ax.plot(t,R/factor,label='R')
  ax.legend()
```

```python
N=120000000
I0=10000
R0=0
S0=N-I0-R0
alpha=0.5
gamma=0.02
t=np.linspace(0,365,365)
y0=S0,I0,R0
```

```python
sol=itg.odeint(SIR,y0,t,args=(alpha,gamma))
```

```python
sol
```

```python
S,I,R=sol.T
```

```python
plotSol(S,I,R,N)
```
