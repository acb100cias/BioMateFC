```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x,r):
  return r*x*(1-x)
```

```python
X=np.arange(0,1,0.01)
plt.plot(X,f(X,0.5))
plt.plot(X,f(X,1))
plt.plot(X,f(X,1.1))
plt.plot(X,f(X,1.5))
plt.plot(X,f(X,1.8))
plt.plot(X,f(X,2))
plt.plot(X,X)
```

```python

for r in [1,1.5,1.8,2,2.3,2.5,2.8,3,3.5,3.8]:
  X=[0.1]
  for i in range(100):
    X.append(X[-1]+f(X[-1],r))
  plt.figure()
  plt.plot(X,'o',label=r)
  plt.legend()
```

```python
X1=np.arange(0,1,0.01)
r=3.94
plt.plot(X1,f(X1,r))
plt.plot(X1,X1)
plt.scatter([0,1-(1/r)],[0,f(1-(1/r),r)])
plt.scatter([0.05],[f(0.05,r)])
xc=0.05
for i in range(500):
    plt.plot([xc,xc],[xc,f(xc,r)], color='k', linestyle='--')
    plt.plot([f(xc,r),xc],[f(xc,r),f(xc,r)], color='k', linestyle='--')
    xc=f(xc,r)
```

```python
n = 10000
r = np.linspace(1, 3.925, n)
iterations = 1000
last = 100
x = 1e-5 * np.ones(n)
lyapunov = np.zeros(n)
fig,ax1  = plt.subplots(1, 1, figsize=(20, 10))
for i in range(iterations):
    x = f(x,r)
    lyapunov += np.log(abs(r - 2 * r * x))
    if i >= (iterations - last):
        ax1.plot(r, x, ',k', alpha=.25)
ax1.set_xlim(1., 3.925)
ax1.set_title('Diagrama de bifurcaciones')
plt.xlabel('r')
plt.ylabel('X')
```
