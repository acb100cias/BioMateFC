```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
def Malthus(x,t,r):
  return r*x
```

```python
T=np.arange(0,10,0.01)
```

```python
X1=odeint(Malthus,0.1,T,args=(0.5,))
```

```python
plt.plot(X1)
```

```python
for i in [0.1,0.5,1.5,2]:
  X=odeint(Malthus,i,T,args=(0.5,))
  plt.plot(X)
```

```python
def logistica(x,t,r,K):
  return(r*x*(1-x/K))
```

```python
T=np.arange(0,50,0.01)
for i in np.arange(0,30,1):
  Y=odeint(logistica,i,T,args=(0.5,20))
  plt.plot(Y)
```

```python
def lcc(x,t,r,K,h):
  return r*x*(1-x/K)-h
```

```python
plt.plot(odeint(lcc,18,T,args=(0.5,20,0.9)))
```

```python
def lctc(x,t,r,K,h):
  return r*x*(1-x/K)-h*x
```
