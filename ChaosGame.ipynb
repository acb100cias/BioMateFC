```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def chaos(start=None,n=None):
  plt.scatter([0,2,4],[0,4,0])
  bounds=np.array([[0,0],[2,4],[4,0]])
  plt.xlim([-0.5,4.5])
  plt.ylim([-0.5,4.5])

  if start==None:
    start=np.array([2,2])
  if n==None:
    n=500
  for i in range(n):
    plt.scatter(start[0],start[1],s=11,c='k')
    dado=np.random.randint(3)
    rnc=np.array(bounds[dado])
    start=start+0.5*(rnc-start)

```

```python
chaos(n=10000)
```
