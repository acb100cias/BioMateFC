```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nw
```

```python
G=nw.karate_club_graph()
```

```python
nw.draw(G,with_labels=True)
```

```python
cg=nw.degree_centrality(G)
ci=nw.betweenness_centrality(G)
cc=nw.closeness_centrality(G)
```

```python
top_grado=sorted(cg,key=cg.get,reverse=True)[:5]
top_inter=sorted(ci,key=ci.get,reverse=True)[:5]
top_cerca=sorted(cc,key=cc.get,reverse=True)[:5]
```

```python
plt.figure(figsize=(10,10))
pos=nw.spring_layout(G)
node_color=['red' if node in top_grado else 'blue' for node in G.nodes()]
nw.draw(G,pos,with_labels=True,node_color=node_color)
```

```python
from networkx.algorithms.community import girvan_newman
```

```python
comp=girvan_newman(G)
primeras_comunidades=next(comp)
```

```python
primeras_comunidades
```

```python
plt.figure(figsize=(10,10))
pos=nw.spring_layout(G)
node_color=['blue' if node in primeras_comunidades[0] else 'orange' for node in G.nodes()]
nw.draw(G,pos,with_labels=True,node_color=node_color)
```

```python
def initialize():
  global g
  g=nw.karate_club_graph()
  g.pos=nw.spring_layout(g)
  for i in g.nodes:
    g.nodes[i]['state']=1 if np.random.random() < 0.1 else 0

def observe():
  global g
  plt.figure(figsize=(10,10))
  nw.draw(g,g.pos,cmap=plt.cm.Wistia,vmin=0,vmax=1,node_color=[g.nodes[i]['state'] for i in g.nodes],with_labels=True)
p_i=0.7
p_r=0.1
def update():
  global g
  a=np.random.choice(list(g.nodes))
  if g.nodes[a]['state'] == 0:
    b=np.random.choice(list(g.neighbors(a)))
    if g.nodes[b]['state'] == 1:
      if np.random.random() > p_i:
        g.nodes[a]['state']=1
  elif np.random.random() > p_r:
      g.nodes[a]['state']=0

```

```python
initialize()
observe()
for i in range(35):
  update()
  observe()
```
