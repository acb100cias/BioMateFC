```python
import numpy as np
import networkx as nw
import matplotlib.pyplot as plt

```

```python
TamPob=500
p=0.01
i=0.01
pi=0.2
pr=0.5
S=0
I=1
```

```python
def initialize():
  global time,network,pos,nextNetwork
  time=0
  network=nw.erdos_renyi_graph(TamPob,p)
  pos=nw.random_layout(network)
  for j in network.nodes:
    if np.random.random() < i:
      network.nodes[j]['state']=I
    else:
      network.nodes[j]['state']=S
  nextNetwork=network.copy()

def observe():
  nw.draw(network,pos,node_color=[network.nodes[k]['state'] for k in network.nodes]
          ,cmap=plt.cm.jet,vmin=0,vmax=1)
  plt.title('time='+str(time))

def update():
  global time,network,nextNetwork
  time+=1
  for k in network.nodes:
    if network.nodes[k]['state']==S:
      nextNetwork.nodes[k]['state']=S
      for j in network.neighbors(k):
        if network.nodes[j]['state']==I:
          if np.random.random() < pi:
            nextNetwork.nodes[k]['state']=I
            break
    else:
      if np.random.random() < pr:
        nextNetwork.nodes[k]['state']=S
      else:
        nextNetwork.nodes[k]['state']=I
  network,nextNetwork=nextNetwork,network

```

```python
initialize()
observe()
for n in range(10):
  update()
  observe()
  plt.show()
```
