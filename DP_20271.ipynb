```python
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt
```

```python
def DP(x,t,a,b,c,d):
  dx0=a*x[0]-b*x[0]*x[1]
  dx1=-c*x[1]+d*x[0]*x[1]
  return np.array([dx0,dx1])
```

```python
T=np.arange(0,500,0.01)
```

```python
sol=odeint(DP,[0.1,0.1],T,args=(0.1,0.3,0.08,0.7))
```

```python
plt.plot(sol)
```

```python
solt=sol.T
```

```python
solt
```

```python
plt.plot(T,solt[0],label='Presas')
plt.plot(T,solt[1],label='Depredadores')
plt.legend()
```

```python
plt.plot(solt[0],solt[1])
plt.title('Plano Fase')
```

```python
for ci in [[0.1,0.1],[0.9,0.1],[0.1,0.9]]:
  sol=odeint(DP,ci,T,args=(0.1,0.3,0.08,0.7))
  solt=sol.T
  plt.plot(solt[0],solt[1],label=str(ci))
plt.title('Plano Fase')
plt.xlabel('Presas')
plt.ylabel('Depredadores')
plt.legend()
```
